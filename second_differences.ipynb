```python
import matplotlib.pyplot as plt
from matplotlib import style
import numpy as np
import sympy as sp
```

```python
x = sp.symbols("x")
display_rad = 1
style.use('ggplot')
```

```python
f = x ** 4

sp.plotting.plot(f, xlim=(-display_rad, display_rad), ylim=(-display_rad, display_rad))
```

```python
df = sp.diff(f, x)
sp.plotting.plot(df, xlim=(-display_rad, display_rad), ylim=(-display_rad, display_rad))
```

```python
d2f = sp.diff(df, x)
sp.plotting.plot(d2f, xlim=(-display_rad, display_rad), ylim=(-display_rad, display_rad))
```

```python
spacing = 0.001
x_num = np.arange(-display_rad, display_rad+spacing, spacing)
```

```python

f_num = x_num ** 4
print(f_num)
fig, ax = plt.subplots()
plt.xlim(-display_rad, display_rad)
plt.ylim(-display_rad, display_rad)
ax.plot(x_num, f_num)
```

```python
df_num = np.append(np.diff(f_num) / spacing, float('inf'))
print(df_num)
fig1, ax1 = plt.subplots()
plt.xlim(-display_rad, display_rad)
plt.ylim(-display_rad, display_rad)
ax1.plot(x_num, df_num)
```

```python
d2f_num = np.append(np.diff(df_num) / spacing, 0)
fig2, ax2 = plt.subplots()
plt.xlim(-display_rad, display_rad)
plt.ylim(-display_rad, display_rad)
ax2.plot(x_num, d2f_num)
```
