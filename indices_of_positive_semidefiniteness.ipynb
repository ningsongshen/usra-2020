```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

```python
x, y = sp.symbols('x y')
H = sp.Matrix([[x, y], [y, x]])
```

```python
eigenvalues = H.eigenvals()
print(eigenvalues)

trace = H[0,0] + H[1,1]
determinant = H.det()

discriminant = (trace ** 2) - (4 * determinant)

lamda_1 = (trace + sp.sqrt(discriminant)) / 2
lamda_2 = (trace - sp.sqrt(discriminant)) / 2

print(lamda_1, lamda_2)

Q, lamda = H.diagonalize()
```
